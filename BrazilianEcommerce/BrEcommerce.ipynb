```python
# 1. Находим месяц первой покупки для каждого уникального клиента
df_delivered['cohort_month'] = df_delivered.groupby('customer_unique_id')['order_purchase_timestamp'].transform('min').dt.to_period('M')

# 2. Вычисляем разницу в месяцах (cohort_index)
# Для этого вычтем из года и месяца текущего заказа год и месяц когорты
def get_cohort_index(df):
    year_diff = df['order_month'].dt.year - df['cohort_month'].dt.year
    month_diff = df['order_month'].dt.month - df['cohort_month'].dt.month
    return year_diff * 12 + month_diff

df_delivered['cohort_index'] = get_cohort_index(df_delivered)

# Посмотрим, что получилось для клиентов с повторными покупками
df_delivered[df_delivered['cohort_index'] > 0][['customer_unique_id', 'order_month', 'cohort_month', 'cohort_index']].head()
```

```python
# Создаем сводную таблицу (pivot table)
cohort_data = df_delivered.pivot_table(index='cohort_month', 
                                       columns='cohort_index', 
                                       values='customer_unique_id', 
                                       aggfunc='nunique')

# Посмотрим на первые несколько строк матрицы
cohort_data.head()
```

```python
# 1. Берем размеры когорт (первая колонка с индексом 0)
cohort_sizes = cohort_data.iloc[:, 0]

# 2. Делим всю таблицу на размеры когорт
retention = cohort_data.divide(cohort_sizes, axis=0)

# 3. Округлим для наглядности
retention.head().round(3)
```

# RFM-анализ (сегментация клиентов)
Нужно рассчитать три метрики для каждого unique_id:
- Recency (R) — как давно была последняя покупка (в днях от «сегодня»).
- Frequency (F) — как часто клиент покупал (количество заказов).
- Monetary (M) — сколько денег он оставил (сумма чеков).

Чтобы  анализ был актуальным в рамках имеющегося набора данных, используем максимальную дату (самый свежий заказ) как точку отсчета. Обычно к ней добавляют один день, чтобы у последнего покупателя показатель Recency не был равен нулю (для удобства расчетов).

```python
print(df_delivered.columns)
df_delivered.head()
```

```python
# 1. Определяем точку отсчета
last_date = df_delivered['order_purchase_timestamp'].max() + pd.Timedelta(days=1)
print(f" точка отсчета {last_date}")
# Агрегируем данные по уникальным ID клиентов
rfm = df_delivered.groupby('customer_unique_id').agg({
    # Разница между точкой отсчета и последней покупкой
    'order_purchase_timestamp': lambda x: (last_date - x.max()).days,
    # Количество уникальных заказов
    'order_id': 'nunique',
    # Сумма всех покупок (так как price уже есть в таблице)
    'price': 'sum'
})

# Переименовываем для удобства
rfm.rename(columns={
    'order_purchase_timestamp': 'Recency',
    'order_id': 'Frequency',
    'price': 'Monetary'
}, inplace=True)

rfm.head()
```

```python
rfm.describe()
```

- Теперь мы точно видим, что делить Frequency на 5 равных групп через pd.qcut нельзя — у первых четырех групп будет одно и то же значение (1 заказ).
Поэтому мы используем комбинированный подход:
R и M делим на 5 групп (квантилей).
F делим вручную на понятные бизнесу сегменты.

```python
# 1. Ранги для Recency (5 - купили совсем недавно, 1 - очень давно)
rfm['R'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1])

# 2. Ранги для Monetary (5 - принесли много денег, 1 - мало)
rfm['M'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5])

# 3. Ранги для Frequency (ручная сегментация)
def f_score(x):
    if x == 1:
        return 1 # Разовые клиенты
    elif x == 2:
        return 2 # Вернулись один раз
    else:
        return 3 # Постоянные (3 и более заказов)

rfm['F'] = rfm['Frequency'].apply(f_score)

# Создаем итоговый RFM-score как строку
rfm['RFM_Score'] = rfm['R'].astype(str) + rfm['F'].astype(str) + rfm['M'].astype(str)

rfm.head()
```

```python
# Словарь для сопоставления R и F с названиями сегментов
segs = {
    r'[1-2][1-2]': 'Hibernating',
    r'[1-2]3': 'At Risk',
    r'3[1-2]': 'About to Sleep',
    r'33': 'Need Attention',
    r'[4-5]1': 'New Customers',
    r'[4-5][2-3]': 'Champions'
}

# Применяем сегментацию на основе R и F
rfm['Segment'] = rfm['R'].astype(str) + rfm['F'].astype(str)
rfm['Segment'] = rfm['Segment'].replace(segs, regex=True)

# Считаем, сколько клиентов в каждом сегменте
segment_counts = rfm['Segment'].value_counts()
print(segment_counts)
```

```python
# Группируем по сегментам и считаем сумму денег
segment_revenue = rfm.groupby('Segment')['Monetary'].sum().sort_values(ascending=False)

print(segment_revenue)
```

```python
# Считаем средний чек по сегментам
segment_avg_check = rfm.groupby('Segment')['Monetary'].mean().sort_values(ascending=False)

print(segment_avg_check)
```

- Champions покупают часто, но их разовые траты скромнее. Их ценность в стабильности и лояльности.
Need Attention и At Risk — это клиенты, которые совершают редкие, но очень дорогие покупки. Потеря даже одного такого клиента ощутима для выручки. Hibernating и New Customers имеют почти одинаковый средний чек (~140-142). Это говорит о том, что поведение новичков пока ничем не отличается от тех, кто в итоге «уснул».

# Заключение по результатам RFM-анализа и EDA
## 1. Текущее состояние бизнеса и динамика
Исследование временных рядов показало стадию активного масштабирования в 2017 году с последующим выходом на «плато» в 2018 году. Стабильность заказов в последние месяцы подтверждает валидность метрики Recency: сегментация отражает актуальное поведение клиентов, а не технические сбои в выгрузке данных. Пиковые нагрузки (ноябрь 2017) указывают на высокую чувствительность аудитории к сезонным промо-акциям.
## 2. Структура клиентской базы и лояльность
Диспропорция сегментов: Большая часть базы представлена «разовыми» покупателями. Это характерно для маркетплейсов, но указывает на низкий уровень Retention (удержания).
- Профиль «Чемпионов»: Наши самые лояльные клиенты ориентированы на товары среднего ценового сегмента (медиана ~90). Это «фундамент» бизнеса, обеспечивающий стабильный денежный поток.
- Аномальные чеки: EDA выявил наличие значительных выбросов в чеках (товары дороже 300-500 при основной массе до 150). Клиенты, совершающие разовые дорогие покупки, попадают в зону риска (Need Attention), так как их поведение нетипично для основной массы.
## 3. Рекомендации по стратегии
- Удержание (Retention): Необходимо внедрить триггерные коммуникации для новичков сразу после первой покупки, чтобы перевести их в категорию постоянных.
- Работа с VIP-выбросами: Клиентов с высоким чеком, но низкой частотой, следует выделить в отдельный поток для качественного анализа (возможно, это B2B-закупки или специфическая ниша товаров).
- Оптимизация ассортимента: Учитывая плотность распределения цен в EDA, основной фокус маркетинга должен быть направлен на товары в диапазоне 50–150, так как они генерируют основной объем транзакций.
